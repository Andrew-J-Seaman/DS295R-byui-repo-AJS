```python
# Import OS and API
import os
from google.colab import userdata

os.environ['GOOGLE_API_KEY'] = userdata.get('GOOGLE_API_KEY')
```

```python
# Import only a portion of the langchain library because it's massive.
from langchain.agents import create_agent

agent = create_agent(
    "google_genai:gemini-3.6-flash"
)
```

```python
# Send first message:
#   - This is a verbose output following a user prompt.
#   - The output includes the natural language reply as well as the meta-data "process" detail dump.
#   - You can customize the output to strip this off details you don't wish to see.
response = agent.invoke({"messages": [{"role": "user", "content": "Hello world!"}]})
```

```python
print(response)
```

```python
# Send second message:
from langchain_core.messages import HumanMessage

result = agent.invoke({'messages': [HumanMessage(content='Hello world!', additional_kwards={}, response_metadata={})]})
```

```python
print(result["messages"][-1].content_blocks)
```
